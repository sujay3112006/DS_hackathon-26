## Objective
Train Random Forest model using Time Series cross-validation, evaluate on test set (Accuracy, Precision, Recall, F1, ROC-AUC), and extract feature importances.

```python
import pandas as pd
import numpy as np
import joblib
import sys
sys.path.append('..')

from src.model_training import train_random_forest_model
from src.evaluation import evaluate_model
from src.feature_engineering import create_features_and_target, time_based_train_test_split

df = pd.read_csv('../data/processed/cleaned_manufacturing_data.csv')
df_encoded, feature_cols, thresh = create_features_and_target(df)
X_tr, X_te, y_tr, y_te, meta, _, _ = time_based_train_test_split(df_encoded, feature_cols)

best_model, artifact = train_random_forest_model(X_tr, y_tr, model_filepath='../models/random_forest_model.pkl')
metrics, cm, fi_df, y_pred = evaluate_model(best_model, X_te, y_te, feature_cols)

print('=== Random Forest Evaluation Metrics ===')
for k, v in metrics.items():
    if k not in ['Confusion_Matrix', 'Classification_Report']:
        print(f'  {k}: {v:.4f}')
```
